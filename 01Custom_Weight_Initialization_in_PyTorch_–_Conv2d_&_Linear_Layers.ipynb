```python
import torch
import torch.nn as nn
import torch.nn.init as init
import math
```

```python
class MyNet(nn.Module):
    def __init__(self):
        super(MyNet, self).__init__()
        self.fc1 = nn.Linear(5, 3)
        self.fc2 = nn.Linear(3, 1)

    def forward(self, x):
        x = torch.relu(self.fc1(x))
        x = self.fc2(x)
        return x
```

```python
model = MyNet()
```

```python
for name, param in model.named_parameters():
      print(f"{name} -> mean: {param.data.mean():.4f}, std: {param.data.std():.4f}")
      print(param, "\n")
```

```python
def check_params(model):
    for name, param in model.named_parameters():
        print(f"{name} -> mean: {param.data.mean():.4f}, std: {param.data.std():.4f}")
        print(param, "\n")
```

```python
x = torch.randn(2, 5)
x
```

## **1. Zero Initialization**

```python
model = MyNet()
for m in model.modules():
    if isinstance(m, nn.Linear):
        init.zeros_(m.weight)
        init.zeros_(m.bias)
```

```python
print("\n--- Zero Initialization ---")
check_params(model)
```

```python
print("Output:", model(x))
```

# **# 2. Constant Initialization**

```python
model = MyNet()
for m in model.modules():
    if isinstance(m, nn.Linear):
        init.constant_(m.weight, 5)
        init.constant_(m.bias, 0)
```

```python
print("\n--- Constant Initialization ---")
check_params(model)
```

```python
print("Output:", model(x))
```

# **Random Initialization (with small values, large values)**

```python
model = MyNet()
for m in model.modules():
    if isinstance(m, nn.Linear):
        init.uniform_(m.weight, -0.01, 0.01)
        init.zeros_(m.bias)

print("\n--- Small Random Uniform ---")
check_params(model)
print("Output:", model(x))
```

```python
model = MyNet()
for m in model.modules():
    if isinstance(m, nn.Linear):
        init.uniform_(m.weight, -1.0, 1.0)
        init.zeros_(m.bias)

print("\n--- Large Random Uniform ---")
check_params(model)
print("Output:", model(x))
```

```python
model = MyNet()
for m in model.modules():
    if isinstance(m, nn.Linear):
        init.normal_(m.weight, mean=0, std=0.01)
        init.zeros_(m.bias)

print("\n--- Small Random Normal ---")
check_params(model)
print("Output:", model(x))
```

```python
model = MyNet()
for m in model.modules():
    if isinstance(m, nn.Linear):
        init.normal_(m.weight, mean=0, std=1.0)
        init.zeros_(m.bias)

print("\n--- Large Random Normal ---")
check_params(model)
print("Output:", model(x))
```

# **Xavier (Glorot) Initialization**

```python
model = MyNet()
for m in model.modules():
    if isinstance(m, nn.Linear):
        init.xavier_uniform_(m.weight)
        init.zeros_(m.bias)

print("\n--- Xavier Uniform ---")
check_params(model)
print("Output:", model(x))
```

```python
model = MyNet()
for m in model.modules():
    if isinstance(m, nn.Linear):
        init.xavier_normal_(m.weight)
        init.zeros_(m.bias)

print("\n--- Xavier Normal ---")
check_params(model)
print("Output:", model(x))
```

# **He Initialization (Kaiming)**

```python
model = MyNet()
for m in model.modules():
    if isinstance(m, nn.Linear):
        init.kaiming_uniform_(m.weight, mode="fan_in", nonlinearity="relu")
        init.zeros_(m.bias)

print("\n--- Kaiming Uniform ---")
check_params(model)
print("Output:", model(x))
```

```python
model = MyNet()
for m in model.modules():
    if isinstance(m, nn.Linear):
        init.kaiming_normal_(m.weight, mode="fan_in", nonlinearity="relu")
        init.zeros_(m.bias)

print("\n--- Kaiming Normal ---")
check_params(model)
print("Output:", model(x))
```

# **LeCun Initialization**

```python
model = MyNet()
for m in model.modules():
    if isinstance(m, nn.Linear):
        fan_in = m.weight.size(1)
        bound = math.sqrt(3.0 / fan_in)
        init.uniform_(m.weight, -bound, bound)
        init.zeros_(m.bias)

print("\n--- LeCun Uniform ---")
check_params(model)
print("Output:", model(x))
```

```python
model = MyNet()
for m in model.modules():
    if isinstance(m, nn.Linear):
        fan_in = m.weight.size(1)
        std = math.sqrt(1.0 / fan_in)
        init.normal_(m.weight, 0.0, std)
        init.zeros_(m.bias)

print("\n--- LeCun Normal ---")
check_params(model)
print("Output:", model(x))
```

# **Uniform Initialization**

```python
model = MyNet()
for m in model.modules():
    if isinstance(m, nn.Linear):
        init.uniform_(m.weight, -0.1, 0.1)
        init.zeros_(m.bias)

print("\n--- Custom Uniform ---")
check_params(model)
print("Output:", model(x))
```

# **Normal (Gaussian) Initialization**

```python
model = MyNet()
for m in model.modules():
    if isinstance(m, nn.Linear):
        init.normal_(m.weight, mean=0, std=0.02)
        init.zeros_(m.bias)

print("\n--- Custom Normal ---")
check_params(model)
print("Output:", model(x))
```

# **one example with conv2d**

```python
class MyModel(nn.Module):
    def __init__(self):
        super(MyModel, self).__init__()
        self.conv1 = nn.Conv2d(3, 8, kernel_size=3, stride=1, padding=1)
        self.fc1 = nn.Linear(8 * 8 * 8, 10)

    def forward(self, x):
        x = torch.relu(self.conv1(x))
        x = torch.flatten(x, 1)
        x = self.fc1(x)
        return x
```

```python
model = MyModel()
```

```python
for name, param in model.named_parameters():
    print(f"{name} -> mean: {param.data.mean():.4f}, std: {param.data.std():.4f}")
```

```python
for name, param in model.named_parameters():
    print(name)
```

# **Custom initialization function**

```python
def initialize_parameters(m):
    if isinstance(m, nn.Conv2d):

        nn.init.kaiming_normal_(m.weight.data,
                                nonlinearity='relu')
        nn.init.constant_(m.bias.data, 0)
    elif isinstance(m, nn.Linear):

        nn.init.xavier_normal_(m.weight.data, gain=nn.init.calculate_gain('relu'))
        nn.init.constant_(m.bias.data, 0)
```

```python
model.apply(initialize_parameters)
```

```python
for name, param in model.named_parameters():
    print(f"{name} -> mean: {param.data.mean():.4f}, std: {param.data.std():.4f}")


```

```python
for name, param in model.named_parameters():
    print(name)
    print(param)
```
